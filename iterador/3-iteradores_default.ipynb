```python
import itertools
```

```python
for i in range(1, 6):
   print(i)
```

```python
frutas = ['maça', 'banana', 'laranja', 'melão', 'limão']
for fruta in frutas:
   print(fruta)
```

```python
pessoa = {'nome':'Alice', 'idade':30}
for chave in pessoa:
   print(chave, pessoa[chave])
```

```python
for indice, fruta in enumerate(frutas):
   print(indice, fruta)
```

```python
nomes = ['Alice', 'bob', 'carlos']
idades = [30, 25, 34]
for nome, idade in zip(nomes, idades):
   print(nome, idade)
```

```python
for i in itertools.count(start=10, step=2):
   if i > 20:
      break
   print(i)
```
